## Selesksi Data Indo timur

```python
print(f'total Data =>',{data['kd_siswa_an'].count()})
data_timur = data[data['wilayah_bagian']=='Indonesia Timur']
print(f'data dari Indonesia Timur',{data_timur['kd_siswa_an'].count()})
data_timur.tail()
```

Grouping Nilai AN

```python
nilai_AN_min = data_timur['nilai_AN'].min()
nilai_AN_max = data_timur['nilai_AN'].max()
interval = (nilai_AN_max - nilai_AN_min) / 3
def klasifikasi_nilai_AN(x):
  if x < nilai_AN_min + interval:
      return 'rendah'
  elif x < nilai_AN_min + 2 * interval:
    return 'cukup'
  else:
    return 'tinggi'

data_timur['kelas_nilai_AN'] = data_timur['nilai_AN'].apply(klasifikasi_nilai_AN)
data_timur[['nilai_AN', 'kelas_nilai_AN']].head()
```
